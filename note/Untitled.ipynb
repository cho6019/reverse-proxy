```python
import random
import time

def gen_r_array_choices(N):
    a = random.choices(range(101), k=N)
    b = random.choices(range(101), k=N)
    return a, b

def check_time(defintion, N):
    s_time = time.time()
    defintion(N)
    e_time = time.time()
    return (e_time-s_time)
```

```python
def gen_r_array_choices_numpy(N):
    a=np.random.choice(range(101), size = N)
    b=np.random.choice(range(101), size = N)
    return a, b
```

```python
num_t=[]
ran_t=[]

def return_time(defintion, list):
    list.extend(check_time(defintion, 10**i) for i in range (1, 7))

return_time(gen_r_array_choices, ran_t)
return_time(gen_r_array_choices_numpy, num_t)
```

```python
for i in range(1, 7):
    start_time = time.time()
    a, b = gen_r_array_choices(10**i)
    end_time = time.time()
    ran_t.append(end_time-start_time)
```

```python
plt.figure(figsize=(7, 9))
x = [1,2,3,4,5,6]
plt.plot(x, num_t, marker='o', linestyle='-', color='b', label="NUMPY")
plt.plot(x, ran_t, marker='s', linestyle='-', color='r', label="RANDOM")
plt.xlabel("count")
plt.ylabel("time")
plt.legend()
plt.show()
```

```python
def gen_r_array_choices_numpy(N):
    a = np.random.choice(range(101), size=N)
    b = np.random.choice(range(101), size=N)
    return a, b

# Python random 기반 함수
def gen_r_array_choices(N):
    a = random.choices(range(101), k=N)
    b = random.choices(range(101), k=N)
    return a, b

def measure_time(func, N):
    result = 0
    for _ in range(10):
        start = time.time()
        func(N)
        end = time.time()
        result += (end - start)
    return result / 10

# 테스트할 N 값 생성 (1부터 1,000,000까지 100개 간격)
N_values = np.linspace(1, 1_000_000, 100, dtype=int)

numpy_times = []
random_times = []
# TODO 각 함수의 실행 시간 계산 위 배열에 추가
for N in N_values:
    numpy_times.append(measure_time(gen_r_array_choices_numpy, N))
    random_times.append(measure_time(gen_r_array_choices, N))


# 선그래프 그리기
plt.figure(figsize=(10, 6)) # 차트 크기 -> 그래프의 가로 크기 10인치, 세로 크기 6인치
plt.xlabel('Array Size') # x축 레이블
plt.ylabel('Time') # y축 레이블
plt.title('Performance : numpy VS random')
plt.grid(True) # 그래프에 그리드 ON
plt.plot(N_values, numpy_times, label='numpy', color='blue')
plt.plot(N_values, random_times, label='random', color='red')
plt.legend() # 범례를 추가
plt.show()
```
