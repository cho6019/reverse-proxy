```python
d_nhh = pd.read_csv("numpy-2.0-1000.csv")
d_nhl = pd.read_csv("numpy-2.0-100.csv")
d_nlh = pd.read_csv("numpy-0.5-1000.csv")
d_nll = pd.read_csv("numpy-0.5-100.csv")
d_phh = pd.read_csv("python-2.0-1000.csv")
d_phl = pd.read_csv("python-2.0-100.csv")
d_plh = pd.read_csv("python-0.5-1000.csv")
d_pll = pd.read_csv("python-0.5-100.csv")
```

```python
def process_percentage_columns(df):
    df["CPU_PERC"] = df["CPU_PERC"].str.replace('%', '').astype(float) / 100
    df["MEM_PERC"] = df["MEM_PERC"].str.replace('%', '').astype(float) / 100

    df.loc[:, "CPU_PERC"] = pd.to_numeric(df["CPU_PERC"], errors="coerce")
    df.loc[:, "MEM_PERC"] = pd.to_numeric(df["MEM_PERC"], errors="coerce")
    return df


d_nhh = process_percentage_columns(d_nhh)
d_nhl = process_percentage_columns(d_nhl)
d_nlh = process_percentage_columns(d_nlh)
d_nll = process_percentage_columns(d_nll)
d_phh = process_percentage_columns(d_phh)
d_phl = process_percentage_columns(d_phl)
d_plh = process_percentage_columns(d_plh)
d_pll = process_percentage_columns(d_pll)
```
